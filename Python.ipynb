### **Conditional Flow Evaluation**

In Python, conditions are evaluated sequentially from top to bottom:
1. If the `if` condition evaluates to `True`, its block is executed, and the rest of the chain is skipped.
2. If it is `False`, the interpreter checks subsequent `elif` statements.
3. If all conditions are `False`, the `else` block (if present) is executed.

*Note: Python uses indentation (usually 4 spaces) to define the block scope instead of curly braces `{}`.*

### **Short-circuit Evaluation in Conditions**

*   **`and` operator:** If the first condition is `False`, Python does not evaluate the second condition (because the result will always be `False`).
*   **`or` operator:** If the first condition is `True`, Python does not evaluate the second condition (because the result will always be `True`).

# **Tuples**

A tuple is a collection which is ordered and **immutable** (cannot be changed after creation). In Python, tuples are written with round brackets `()`.

### Key Characteristics:
*   **Ordered:** Elements have a defined order that will not change.
*   **Immutable:** You cannot add, remove, or modify elements once the tuple is created.
*   **Allow Duplicates:** Can contain multiple identical elements.
*   **Usage in DSA:** Excellent for representing coordinates `(x, y)`, RGB values, or as dictionary keys (since lists cannot be keys because they are mutable/unhashable).


````python
# Creating Tuples
point = (10, 20)
print('Point:', point)

# Tuple Unpacking
x, y = point
print(f'Unpacked values: x={x}, y={y}')

# Single element tuple must have a trailing comma
single_val = (5,)
print('Type of (5,):', type(single_val))
not_a_tuple = (5)
print('Type of (5):', type(not_a_tuple))

# Common tuple methods: count() and index()
my_tuple = (1, 2, 3, 2, 4, 2)
print('Count of 2:', my_tuple.count(2))
print('Index of 3:', my_tuple.index(3))

# Accessing elements is identical to lists (slicing, indexing)
print('Slice of tuple:', my_tuple[1:4])
````

# **Sets**

A set is a collection which is **unordered**, **mutable** (though elements must be immutable/hashable), and **unindexed**. In Python, sets are written with curly brackets `{}`.

### Key Characteristics:
*   **Unique Elements:** No duplicate values are allowed.
*   **Unordered:** No index or guaranteed order.
*   **O(1) Lookup Time:** Behind the scenes, sets are implemented as hash tables. Checking if an element exists is extremely fast: $O(1)$ on average.
*   **Usage in DSA:** Crucial for tracking visited nodes in Graph/BFS/DFS algorithms, removing duplicates from lists, and checking memberships.


````python
# Creating sets
visited = {'A', 'B', 'C'}
print('Initial set:', visited)

# O(1) membership check (in operator)
print("'A' in visited:", 'A' in visited)
print("'D' in visited:", 'D' in visited)

# Adding and removing elements
visited.add('D')
visited.add('A') # No effect, duplicate
print('After adding:', visited)

visited.remove('B') # Raises KeyError if not found
visited.discard('Z') # Safe remove: does not raise error if not found
print('After removals:', visited)

# Set Operations (Union, Intersection, Difference)
set_a = {1, 2, 3, 4}
set_b = {3, 4, 5, 6}

print('Union (|):', set_a | set_b)
print('Intersection (&):', set_a & set_b)
print('Difference (-):', set_a - set_b)
print('Symmetric Difference (^):', set_a ^ set_b)

# Set Comprehension
even_squares = {x**2 for x in range(10) if x % 2 == 0}
print('Set Comprehension:', even_squares)
````

# **Shallow Copy vs Deep Copy**

In Python, assigning a list/dict to a new variable just copies the **reference**, not the actual object. If you modify the new variable, the original also changes!

*   **Shallow Copy (`copy()` or `[:]`):** Copies the outer object, but nested objects (like matrices/nested lists) are still shared by reference.
*   **Deep Copy (`copy.deepcopy()`):** Recursively copies the outer object and all nested objects. Totally independent.
*   **Usage in DSA:** Deep copy is critical when backtracking in recursive algorithms (e.g., saving grid states, permutation lists).


````python
import copy

# 1. Assignment (Shared reference)
original = [[1, 2], [3, 4]]
shared = original
shared[0][0] = 99
print('Original after modifying shared assignment:', original)

# 2. Shallow Copy
original = [[1, 2], [3, 4]]
shallow = original.copy() # or original[:]
shallow[0][0] = 99 # Changes nested element in both
shallow.append([5, 6]) # Only changes shallow list outer layer
print('Original after modifying nested in shallow copy:', original)
print('Shallow after outer append:', shallow)

# 3. Deep Copy
original = [[1, 2], [3, 4]]
deep = copy.deepcopy(original)
deep[0][0] = 99 # Totally isolated
print('Original after modifying deep copy:', original)
print('Deep copy:', deep)
````

# **Built-in Functions for Iterables**

Python provides several highly optimized built-in functions that make operations on sequences elegant and fast.

*   **`zip()`**: Parallel iteration over multiple sequences.
*   **`sorted()`**: Returns a new sorted list (keeps original intact, unlike `list.sort()`).
*   **`any()`**: Returns `True` if at least one element is `True`.
*   **`all()`**: Returns `True` if all elements are `True`.


````python
# 1. zip()
names = ['Alice', 'Bob', 'Charlie']
scores = [85, 92, 78]
for name, score in zip(names, scores):
    print(f'{name} scored {score}')

# 2. sorted() vs list.sort()
nums = [4, 1, 8, 2]
new_sorted = sorted(nums, reverse=True)
print('Original remains unsorted:', nums)
print('New sorted list:', new_sorted)

# 3. any() & all()
booleans = [True, False, True]
print('any():', any(booleans)) # At least one is True
print('all():', all(booleans)) # Not all are True

# Checking conditions using list comprehensions and any/all
print('Are all numbers positive?', all(x > 0 for x in [1, 2, -3, 4]))
print('Is there any even number?', any(x % 2 == 0 for x in [1, 3, 5, 6]))
````

# **Generators (`yield`)**

A Generator in Python is a function that returns an iterator using the `yield` keyword. Instead of returning a fully formed list in memory, it produces values one at a time on demand.

*   **Memory Efficiency:** Crucial for handling massive sequences or infinite loops where storing all values in memory would cause an out-of-memory error.


````python
# Generator function
def fibonacci_generator(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

# Using the generator
fib_gen = fibonacci_generator(5)
print('Generator object:', fib_gen)
for val in fib_gen:
    print(val)

# Generator expression (analogous to list comprehension but lazy)
squares_gen = (x**2 for x in range(1000000))
print('First square:', next(squares_gen))
print('Second square:', next(squares_gen))
````

````python
#  Global Variable
x = 10  # global
def my_func():
    print(x)  # Accessing global variable

my_func()       # Output: 10
print(x)        # Output: 10

# Modifying Global Variable Without global Keyword
x = 10

def my_func():
    x = x + 1  #  ERROR: local variable referenced before assignment
    print(x)

my_func()
# Why this fails: Python thinks you're creating a new local x, but you try to read it before assigning.
````

````python
# Correct Way: Use global
x = 10
def my_func():
    global x
    x = x + 1
    print(x)
my_func()       # Output: 11
print(x)        # Output: 11
# Now x is explicitly marked as global, so changes affect the global variable.
````


````python
# What is a @staticmethod?
'''
A static method is a method inside a class that:
- Does NOT take self or cls as the first parameter
- Cannot access or modify instance (self) or class (cls) data
- Belongs to the class, not to any specific object
- Can be called using either the class name or an instance
'''

#  When to Use It?
'''
Use @staticmethod when the method:
Logically belongs to the class
Doesn’t need to access or modify any instance or class variables
Is like a utility/helper function'''
````

````python
class MathOperations:

    @staticmethod
    def add(x, y):
        return x + y

# Call using class name
print(MathOperations.add(5, 3))  # Output: 8

# Or using an instance
obj = MathOperations()
print(obj.add(10, 7))  # Output: 17
````

````python
# What is a @classmethod?
'''
A class method:
Takes cls (not self) as its first argument
Can access and modify class-level attributes
Belongs to the class, not an instance
Can be called using either the class name or an object
'''

# When to Use It?
'''
Use @classmethod when:
You want to access or modify class variables
You want to create alternative constructors (e.g., from_string())
'''
````

````python
# ----------------------------------------
# Multiple Inheritance: Company & Country
# ----------------------------------------

# Parent class 1
class company:
    def __init__(self, com_name):
        self.com_name = com_name  # Company name

    def company_info(self):
        print(f"Company name is {self.com_name}")

# Parent class 2
class country:
    def __init__(self, country_name):
        self.country_name = country_name  # Country name

    def country_info(self):
        print(f"Country name is {self.country_name}")

# Child class inheriting from both company and country
class employee(company, country):
    def __init__(self, com_name, emp_name, country_name):
        self.emp_name = emp_name  # Employee name

        # Initialize both parent classes
        company.__init__(self, com_name)
        country.__init__(self, country_name)

    def full_info(self):
        # Show employee info including company and country
        print(f"The employee {self.emp_name} lives in {self.country_name} and works for {self.com_name}")

    def all_info_child(self):
        print("This is running from employee class")

        # Call parent methods explicitly
        company.company_info(self)
        country.country_info(self)

# -----------------------
# Create object & call methods
# -----------------------
emp1 = employee("Celebal", "Rahul", "India")

emp1.company_info()       # From company
emp1.full_info()          # From employee
emp1.all_info_child()     # Uses both company and country methods
````

````python
# --------------------------------------------
# Multilevel Inheritance: Company → Department → Employee
# --------------------------------------------

# Base class (Level 1)
class Company:
    def __init__(self, com_name):
        self.com_name = com_name  # Company name

    def company_info(self):
        print(f"Company Name: {self.com_name}")

# Derived class from Company (Level 2)
class Department(Company):
    def __init__(self, com_name, dept_name):
        super().__init__(com_name)         # Initialize Company
        self.dept_name = dept_name         # Department name

    def department_info(self):
        print(f"Department Name: {self.dept_name}")

# Derived class from Department (Level 3)
class Employee(Department):
    def __init__(self, com_name, dept_name, emp_name):
        super().__init__(com_name, dept_name)  # Initialize Department (which calls Company)
        self.emp_name = emp_name               # Employee name

    def employee_info(self):
        print(f"Employee Name: {self.emp_name}")

    def full_info(self):
        print("\n--- Full Employee Information ---")
        self.employee_info()
        self.department_info()
        self.company_info()

# --------------------------------------------
# Object creation and method calls
# --------------------------------------------

emp1 = Employee("Celebal", "AI/ML", "Rahul")

emp1.employee_info()      # From Employee
emp1.department_info()    # From Department
emp1.company_info()       # From Company

emp1.full_info()          # All info combined
````

# **Dunder / Magic Methods**

Dunder methods (starting and ending with double underscores `__`) allow you to define behavior for built-in operations on custom objects. For instance:
*   `__str__`: User-friendly string representation (`print(obj)`).
*   `__repr__`: Official/developer-friendly string representation.
*   `__len__`: Custom length logic (`len(obj)`).
*   `__eq__`: Comparison logic (`obj1 == obj2`).


````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None
        
    def __str__(self):
        return f'Node({self.value})'
        
    def __repr__(self):
        return f'Node(value={self.value}, next={self.next})'

# Instantiating Node
node_a = Node(10)
node_b = Node(20)
node_a.next = node_b

print('String representation (__str__):', node_a)
print('Representation representation (__repr__):', repr(node_a))
````

````python
from collections import deque

# Create a deque
dq = deque([1, 2, 3])

# Append to the right (like list.append)
dq.append(4)

# Append to the left
dq.appendleft(0)

print(dq)  # deque([0, 1, 2, 3, 4])

# Pop from the right
dq.pop()  # 4

# Pop from the left
dq.popleft()  # 0

print(dq)  # deque([1, 2, 3])
````

# **heapq (Min-Heap)**

Python's `heapq` module provides an implementation of the Min-Heap algorithm (Priority Queue). 

### Key Characteristics:
*   **Min-Heap:** The smallest element is always at the root (`heap[0]`).
*   **Time Complexities:**
*   **Heapify (convert list to heap):** $O(N)$
*   **Push (insert element):** $O(\log N)$
*   **Pop (extract minimum element):** $O(\log N)$
*   **Max-Heap in Python:** Since `heapq` only supports min-heaps, we implement max-heaps by pushing negative values (e.g. `-val`), and negating them again when popped.


````python
import heapq

# 1. Min-Heap Example
nums = [5, 7, 9, 1, 3]
heapq.heapify(nums) # Convert list to heap in-place: O(N)
print('Heapified list (Note order):', nums)

# Push element: O(log N)
heapq.heappush(nums, 2)
print('Heap after push(2):', nums)

# Pop minimum element: O(log N)
min_val = heapq.heappop(nums)
print('Popped min value:', min_val)
print('Heap after pop:', nums)

# 2. Max-Heap Example (Push negative values)
max_heap = []
for val in [5, 7, 9, 1, 3]:
    heapq.heappush(max_heap, -val)

print('Popped max value (negated back):', -heapq.heappop(max_heap))
````

# **bisect (Binary Search)**

The `bisect` module provides support for maintaining a list in sorted order without having to sort the list after each insertion. It uses standard binary search under the hood.

*   `bisect_left(arr, x)`: Returns the index where element `x` should be inserted to maintain order. If `x` already exists, returns the index of the first occurrence.
*   `bisect_right(arr, x)`: Same, but if `x` already exists, returns the index after the last occurrence.
*   `insort(arr, x)`: Inserts `x` in sorted order: $O(N)$ due to list insertion shift, but search is $O(\log N)$.


````python
import bisect

# Sorted list
sorted_arr = [1, 2, 4, 4, 5, 7, 9]

# Find insertion positions (binary search)
idx_left = bisect.bisect_left(sorted_arr, 4)
idx_right = bisect.bisect_right(sorted_arr, 4)

print('bisect_left(4) index:', idx_left)   # Index 2
print('bisect_right(4) index:', idx_right) # Index 4

# Maintain sorted order
bisect.insort(sorted_arr, 6)
print('After insort(6):', sorted_arr)
````

````python
# Itertools
from itertools import product, permutations, combinations, accumulate, groupby, count, cycle, repeat
import operator

# 1. product: Cartesian product of input iterables
print("Product:")
for p in product([1, 2], ['a', 'b']):
    print(p)
print()

# 2. permutations: all ordered arrangements of length 2
print("Permutations:")
for p in permutations([1, 2, 3], 2):
    print(p)
print()

# 3. combinations: all unordered selections of length 2
print("Combinations:")
for c in combinations([1, 2, 3], 2):
    print(c)
print()

# 4. accumulate: cumulative sums (default) and products (using operator.mul)
print("Accumulate (sum):")
print(list(accumulate([1, 2, 3, 4])))

print("Accumulate (product):")
print(list(accumulate([1, 2, 3, 4], operator.mul)))
print()

# 5. groupby: groups consecutive items by key
data = [('a', 1), ('a', 2), ('b', 3), ('b', 4), ('a', 5)]
print("Groupby:")
for key, group in groupby(data, key=lambda x: x[0]):
    print(key, list(group))
print()

# 6. Infinite iterators examples (only print a few to avoid infinite loop)

print("Count from 10 by 2:")
for i in count(10, 2):
    if i > 20:
        break
    print(i)
print()

print("Cycle through ['A', 'B', 'C'] (6 items):")
cycle_iter = cycle(['A', 'B', 'C'])
for _ in range(6):
    print(next(cycle_iter))
print()

print("Repeat 'Hello' 3 times:")
for item in repeat('Hello', 3):
    print(item)
````
