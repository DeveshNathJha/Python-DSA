# Find Armstrong Number

```python
def is_armstrong(num_str):
    n = len(num_str)
    total = 0  # Fix 1: Initialize total
    
    for i in num_str:
        # Fix 2 & 3: Raise digit to power of n (number of digits)
        total += int(i)**n 
        
    if total == int(num_str):
        return True
    else:
        return False # Fix 4: Add return keyword

number = input("Enter a number: ")
if is_armstrong(number): # You don't need "== True", if is enough
    print(f"{number} is an Armstrong number")
else:
    print(f"{number} is NOT an Armstrong number")
```

```python
# Pythonic version...........
def is_armstrong(num_str):
    n = len(num_str)
    total = sum(int(digit)**n for digit in num_str)
    return total == int(num_str)
```

```python
# With TRY-EXCEPT
def is_armstrong(number_str):
    # n = total number of digits
    n = len(number_str)
    
    # Use 'total' instead of 'sum' (Built-in shadowing avoidance)
    total = 0
    
    # Try-Except to handle non-numeric inputs
    try:
        num_val = int(number_str)
        
        for digit in number_str:
            total += int(digit) ** n
            
        return total == num_val
    except ValueError:
        return False

# Testing
user_input = input("Enter a number: ").strip()
if is_armstrong(user_input):
    print(f"{user_input} is an Armstrong number 🚀")
else:
    print(f"{user_input} is not an Armstrong number")
```

# Calculate the sum of digits of a number

```python
# Function to calculate the SUM of digits using a while loop
def sum_of_digits(number):
    total = 0
    while number > 0:
        # number % 10 gets the last digit (e.g., 123 % 10 = 3)
        total += number % 10 
        # number // 10 removes the last digit (e.g., 123 // 10 = 12)
        number = number // 10
    return total

# Function to calculate the SUM of digits using string conversion (Pythonic way)
def sum_of_digits_str(number_str):
    # Convert each character back to int and sum them up
    return sum(int(digit) for digit in number_str)

# Testing
num = int(input("Enter number for sum: "))
print(f"Sum of digits: {sum_of_digits(num)}")
```

```python
# More Cleaner Version
def count_each_word(input_string):
    words = input_string.lower().split()
    freq = {}
    
    for word in words:
        # .get(word, 0) ka matlab hai: agar word na mile toh 0 maan lo
        freq[word] = freq.get(word, 0) + 1
        
    return {key:value for key,value in freq.items() if value > 1}
```

# Using Backtracking

```python
def find_permutations(s, start, end):
    # Base Case: Jab hum last character tak pahunch jayein
    if start == end:
        print("".join(s))
        return

    for i in range(start, end):
        # 1. SWAP: Current index ko start position se badal do
        s[start], s[i] = s[i], s[start]

        # 2. RECURSE: Agle position ke liye function call karo
        find_permutations(s, start + 1, end)

        # 3. BACKTRACK (Swap Back): Original state wapas lao
        # Taaki agle iteration mein naya swap sahi ho sake
        s[start], s[i] = s[i], s[start]

# Chalaane ka tareeka:
string_list = list("abc")
find_permutations(string_list, 0, len(string_list))
```

# Handle duplicate characters ("aab")

```python
def find_permutations(s):
    # base case: only one char -> single permutation
    if len(s) == 1:
        return [s]

    res = []

    # sort so duplicates come together (important for skipping)
    s = "".join(sorted(s))

    for i in range(len(s)):

        # skip duplicates:
        # if current char same as previous, skip this branch
        # prevents generating same permutation again
        if i > 0 and s[i] == s[i-1]:
            continue

        # fix one character
        fix = s[i]

        # remaining string after removing fixed char
        remaining = s[:i] + s[i+1:]

        # recursively find permutations of remaining
        for p in find_permutations(remaining):
            # attach fixed char in front
            res.append(fix + p)

    return res 


print(find_permutations("aab"))
```

# Why defaultdict(int)?

It automatically initializes missing keys to 0.

```python
# Using Default Dict
from collections import defaultdict

def check(s,p):
    freq1 = defaultdict(int)
    freq2 = defaultdict(int)

    for i in s:
        freq1[i] += 1

    for j in p:
        freq2[j] += 1

    return freq1 == freq2

print(check("Hello","olleH"))        
```

# Count Vowels and Consonants in a given string

```python
def count_(inp):
    s = {"a","e","i","o","u"}
    countV = 0
    countC = 0
    for i in inp:
        if i.isalpha():
            if i.lower() in s:
                countV += 1
            else:
                countC += 1   

        else:
            continue
        
    return [countV,countC]    

print(count_("Hello Everyone"))        
```

```python
# More Pythonic
def count_(s):
    vowels = "aeiou"
    v = sum(1 for c in s.lower() if c.isalpha() and c in vowels)
    c = sum(1 for c in s.lower() if c.isalpha() and c not in vowels)
    return [v,c]

print(count_("Hello Sir! All Good?"))
```

# Program to gives Output: a2b2c3d2 for the Input String Str = “aabbcccdd”

```python
def compress_string(s):
    if not s:
        return "" # Return empty string if the string is empty

    res = ""
    count = 1

    for i in range(1, len(s)):
        # Agar current char pichle char ke barabar hai
        if s[i] == s[i-1]:
            count += 1
        else:
            # Jaise hi naya char mile, purane ka result add kar do
            res += s[i-1] + str(count)
            count = 1 # Count reset kar do naye char ke liye

    # Last waale group ko add karna mat bhoolna
    res += s[-1] + str(count)
    
    return res

print(compress_string("aabbcccdd")) # Output: a2b2c3d2
```

# Longest Substring Without Repeating Characters
Problem

### Given a string s, find the length of the longest substring without repeating characters.

- A substring is a contiguous sequence of characters within a string.

Return the length of the longest such substring.

Examples
Input: s = "abcabcbb"
Output: 3
Explanation: The answer is "abc", with length 3.

```python
def len_longest(inp_str):
    s = set()
    start = 0
    max_len = 0

    for end in range(len(inp_str)):

        while inp_str[end] in s:
            s.remove(inp_str[start])
            start += 1

        s.add(inp_str[end])    
        max_len = max(max_len, end-start+1)
    return max_len

print(len_longest("abcabcbb"))            
```

# Return longest substring

```python
def longest_substring(s):
    seen = set()      # Ye current window ke unique characters ko yaad rakhta hai
    start = 0         # Window ki shuruat (Left pointer)
    max_len = 0       # Sabse lambi substring ki length
    max_start = 0     # Sabse lambi substring kahan se shuru hui (result nikalne ke liye)

    # 'end' hai Window ka khatma (Right pointer) jo aage badhta rahega
    for end in range(len(s)):

        # Agar current character pehle se 'seen' mein hai (Duplicate mil gaya!)
        while s[end] in seen:
            # Window ko peeche (start) se chhota karo
            # Aur jo characters bahar ja rahe hain unhe set se hatao
            seen.remove(s[start])
            start += 1

        # Naya character set mein daal do (Window badi ho gayi)
        seen.add(s[end])

        # Kya ye nayi window purani wali se lambi hai?
        if end - start + 1 > max_len:
            max_len = end - start + 1  # Nayi max length update karo
            max_start = start          # Uska starting point yaad rakho

    # Final result slice karke return karo
    return s[max_start:max_start + max_len]

print(longest_substring("abcabcbb")) # Output: "abc"
```

# Find common elements between two arrays

```python
def common(arr1, arr2):
    s2 = set(arr2)  # O(m) to create the set
    result = set()
    for i in arr1:
        if i in s2: # O(1) lookup
            result.add(i)
    return result


arr1 = [6,7,2,3,4,5]
arr2 = [2,3,4,5,6,7]

print(common(arr1,arr2))


```

# Sort an array without using in-built method

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False # Check karne ke liye ki swap hua ya nahi
        
        # n-1-i tak hi check karenge kyunki aakhri i elements sorted hain
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                # Swap numbers
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        
        # Agar koi swap nahi hua, toh loop se bahar nikal jao
        if not swapped:
            break
            
    return arr

arr = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
print(bubble_sort(arr))
```

```python
def insertion_sort(arr):
    n = len(arr)

    # Index 1 se start karte hain kyunki pehla element (index 0) pehle se sorted maana jata hai
    for i in range(1, n):
        key = arr[i]  # Woh element jise humein sahi jagah "insert" karna hai

        j = i - 1     # Key ke left side wale saare elements ko check karne ke liye index

        # Jab tak j zero se bada hai AUR peeche wala element key se bada hai...
        while j >= 0 and key < arr[j]:
            # Peeche wale bade element ko ek position right shift kar do
            arr[j + 1] = arr[j]
            j -= 1    # Aur peeche check karne ke liye j ko kam karo
        
        # Sahi jagah milte hi, key ko wahan rakh do
        # (j + 1 isliye kyunki loop khatam hone par j ek step extra peeche chala jata hai)
        arr[j + 1] = key   
        
    return arr     

# Testing
num = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
print(insertion_sort(num))        
```

```python
def selection_sort(arr):
    n = len(arr)
    for i in range(n-1):
        mini = i

        for j in range(i+1, n-1):
            if arr[j] < arr[i]:
                mini = j
                arr[mini], arr[i] == arr[i], arr[mini]
    return arr


# Testing
num = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
print(selection_sort(num))  
```

# Merge Sort

```python
def merge_array(arr1, arr2):
    # Pointers to track our position in both arrays
    i, j = 0, 0
    m, n = len(arr1), len(arr2)
    res = []
    
    # 1. Main Loop: Compare elements from both arrays
    # Remember: Use 'and' because we need items in BOTH arrays to compare
    while i < m and j < n:
        if arr1[i] <= arr2[j]:
            res.append(arr1[i])
            i += 1
        else:    
            res.append(arr2[j])
            j += 1
            
    # 2. Cleanup: One array will finish first. 
    # These loops pick up whatever is left over.
    
    # If arr1 still has elements left
    if i < m:
        while i < m:
            res.append(arr1[i])
            i += 1

    # If arr2 still has elements left
    if j < n:
        while j < n:
            res.append(arr2[j])
            j += 1

    return res 

def merge_sort(arr):   
    l = len(arr)
    
    # 3. Base Case: Very important! 
    # If the list has only 1 item (or 0), it is already sorted. Stop here.
    if l <= 1:
        return arr
    
    # 4. Splitting: Find the middle point
    # Use // for integer division (e.g., 5 // 2 = 2)
    mid =  l // 2
    left_arr = arr[:mid]   # Elements from start up to mid
    right_arr = arr[mid:]  # Elements from mid to the end

    # 5. Recursion: Keep splitting until we hit the base case
    left = merge_sort(left_arr)
    right = merge_sort(right_arr)

    # 6. Merge: Combine the sorted halves back together
    return merge_array(left, right)

# Testing the code
num = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
print("Original:", num)
print("Sorted:  ", merge_sort(num))      
```

# Quick Sort

```python
def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr)//2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left)+middle+quick_sort(right)

ex = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
sorted_ex = quick_sort(ex)
print(sorted_ex)     
```

```python
def partition(nums, low, high):
    pivot = nums[low]  # Taking the first element as pivot
    i = low + 1        # Start 'i' after the pivot
    j = high           # Start 'j' at the end

    while True:
        # Move 'i' to the right as long as it's smaller than the pivot
        while i <= j and nums[i] <= pivot:
            i += 1
        # Move 'j' to the left as long as it's larger than the pivot
        while i <= j and nums[j] >= pivot:
            j -= 1

        if i <= j:
            # If they haven't crossed, swap the "out of place" elements
            nums[i], nums[j] = nums[j], nums[i]
        else:
            # If they crossed, we are done
            break

    # Swap the pivot (at 'low') with 'j' (the last element smaller than pivot)
    nums[low], nums[j] = nums[j], nums[low]
    return j

def quick_sort(nums, low, high):
    if low < high:
        p = partition(nums, low, high)
        quick_sort(nums, low, p - 1)
        quick_sort(nums, p + 1, high)

# Testing it correctly
ex = [1, 2, 3, 2, 3, 1, 5, 3, 9, 86, 42, 1]
quick_sort(ex, 0, len(ex) - 1)
print(ex) 
```

```python
# If Pivot is HIGH
def quick_sort_inplace(arr, low, high):
    if low < high:
        pivot_index = partition(arr, low, high)
        quick_sort_inplace(arr, low, pivot_index - 1)
        quick_sort_inplace(arr, pivot_index + 1, high)


def partition(arr, low, high):
    pivot = arr[high]  # last element as pivot
    i = low - 1
    
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


# Example
data = [10, 7, 8, 9, 1, 5]
quick_sort_inplace(data, 0, len(data) - 1)
print(data)
```

# Find the missing number in an Array if it starts from 1

```python
def find_missing(arr):
    # n should be the total numbers expected (length + 1 missing number)
    n = len(arr) + 1 
    
    # Use // for integer division to avoid float output (like 4.0)
    expected_sum = (n * (n + 1)) // 2
    
    # sum(arr) is faster than using a generator 'sum(i for i in arr)'
    actual_sum = sum(arr)
    
    return expected_sum - actual_sum

arr = [1, 2, 3, 5, 6, 7, 8]
print(f"Missing number is: {find_missing(arr)}") 
# Output: 4
```

# Agar list [10, 11, 13] jaisi hai, toh n(n+1)/2 wala formula kaam nahi karega, kyunki woh hamesha 1 se start hone wale numbers ke liye hota hai.

## Iske liye aapko Arithmetic Progression (AP) ka Sum wala formula use karna padega.

Formula:
### Sum = $\frac{n}{2} \times (\text{First Number} + \text{Last Number})$

Jahan:

n: Total numbers kitne hone chahiye (len(arr) + 1)
First Number: List ka sabse chota number (min(arr))
Last Number: List ka sabse bada number (max(arr))

```python
def find_missing_any_range(arr):
    if not arr:
        return None
    
    n = len(arr) + 1
    first = min(arr)
    last = max(arr)
    
    # AP Sum Formula: (n/2) * (a + l)
    expected_sum = (n * (first + last)) // 2
    actual_sum = sum(arr)
    
    return expected_sum - actual_sum

arr = [10, 11, 13]
print(f"Missing number is: {find_missing_any_range(arr)}")
# Output: 12
```

```python
def find_missing_xor(arr):
    first = min(arr)
    last = max(arr)
    
    xor_all = 0
    
    # Step 1: XOR all numbers that SHOULD be there
    for i in range(first, last + 1):
        xor_all ^= i
        
    # Step 2: XOR all numbers that ARE actually there
    for i in arr:
        xor_all ^= i
        
    # Jo bacha, wahi missing number hai!
    return xor_all

arr = [10, 11, 13]
print(f"Missing number (using XOR): {find_missing_xor(arr)}")
```

# Array consists of integers and special characters,sum only integers

```python
def find_sum(arr):
    res = 0
    for i in arr:
        # Check if the item is an integer or a float
        if isinstance(i, (int, float)):
            res += i 
    return res 

arr = [1,2,3,'a','e',43,23,'h',8]
print(f"Sum is : {find_sum(arr)}")

```

### isinstance(i, (int, float)): This checks if the element i is a number. This is safer than checking for strings because it directly identifies what you want to add.

### Avoids Errors: It prevents the code from trying to call a string method (like .isalpha()) on an integer, which would cause the program to crash.

### Result: For your input, the sum will be 80 ($1+2+3+43+23+8$).

```python
# Alternet one liner
def find_sum(arr):
    return sum(i for i in arr if isinstance(i, (int, float)))
```

# Count Odd and Even number from given array

```python
def count_odd_even(arr):
    even = 0
    odd = 0

    for i in arr:
        # Check kar rahe hain ki element number hai ya nahi
        if isinstance(i, int): 
            if i % 2 == 0:
                even += 1
            else:
                odd += 1
        else:
            # Agar string wagera hai to skip kar do
            continue

    return even, odd

arr = [1, 2, 3, 4, 5, 'apple', 6, 7, 8, 9] # Ab ye crash nahi hoga
even, odd = count_odd_even(arr) 
print(f"Even: {even}, Odd: {odd}")
```

# 1. Maximum Number of Vowels in a Substring of Size K

Problem
Given string s and integer k, return max number of vowels in any substring of length k.

Example:

Input: s = "abciiidef", k = 3
Output: 3
Explanation: "iii"


```python
VOWELS = set("aeiou") # Only need lowercase because we use .lower() below

def no_vowels(s, k):
    count = 0
    max_count = 0

    for i in range(len(s)):
        # 1. Add the current character to the window
        # We convert it to lowercase to match our VOWELS set
        if s[i].lower() in VOWELS:
            count += 1
        
        # 2. Remove the character that is now outside the window (left side)
        if i >= k:
            if s[i - k].lower() in VOWELS:
                count -= 1

        # 3. Track the maximum vowels seen in any window
        max_count = max(count, max_count)    
        
        # 4. If we found a window with k vowels, we can't do better. Exit early!
        if max_count == k:
            return k

    return max_count        

# Test cases
print(no_vowels("abciiidef", 3)) # Output: 3
print(no_vowels("aeiou", 2))     # Output: 2
print(no_vowels("leetcode", 3))  # Output: 2
```

# 2. Length of Longest Substring Without Repeating Characters


Input: "abcabcbb"
Output: 3

```python
def len_longest(inp_str):
    s = set()
    start = 0
    max_len = 0

    for end in range(len(inp_str)):

        while inp_str[end] in s:
            s.remove(inp_str[start])
            start += 1

        s.add(inp_str[end])    
        max_len = max(max_len, end-start+1)
    return max_len

print(len_longest("abcabcbb"))            
```

# Return Longest Substring Without Repeating Characters

```python
def longest_substring(s):
    seen = set()      # Ye current window ke unique characters ko yaad rakhta hai
    start = 0         # Window ki shuruat (Left pointer)
    max_len = 0       # Sabse lambi substring ki length
    max_start = 0     # Sabse lambi substring kahan se shuru hui (result nikalne ke liye)

    # 'end' hai Window ka khatma (Right pointer) jo aage badhta rahega
    for end in range(len(s)):

        # Agar current character pehle se 'seen' mein hai (Duplicate mil gaya!)
        while s[end] in seen:
            # Window ko peeche (start) se chhota karo
            # Aur jo characters bahar ja rahe hain unhe set se hatao
            seen.remove(s[start])
            start += 1

        # Naya character set mein daal do (Window badi ho gayi)
        seen.add(s[end])

        # Kya ye nayi window purani wali se lambi hai?
        if end - start + 1 > max_len:
            max_len = end - start + 1  # Nayi max length update karo
            max_start = start          # Uska starting point yaad rakho

    # Final result slice karke return karo
    return s[max_start:max_start + max_len]

print(longest_substring("abcabcbb")) # Output: "abc"
```

# 3. Find All Substrings of Size K With Distinct Characters

Return count of substrings with no duplicates

Input: s = "havefunonleetcode", k = 5
Output: 6

```python
def sub_strings(s, k):
    if k >= len(s) or k <= 0:
        return 0
    
    res = []

    count = 0

    # Iterate through each possible window of size K
    for i in range(len(s)-k+1):
        window = s[i : i+k]

        if len(set(window)) == k:
            count += 1
            res.append(s[i : i + k])
    return res    



s = "havefunonleetcode"
sub_strings(s, 5)              
```

```python
def sub_strings(s, k):
    if k > len(s) or k <= 0: return []
    
    res = []
    char_count = {}
    left = 0
    
    for right in range(len(s)):
        # Add character to window
        char_count[s[right]] = char_count.get(s[right], 0) + 1
        
        # Shrink window if it exceeds size k
        if right - left + 1 > k:
            char_count[s[left]] -= 1
            if char_count[s[left]] == 0:
                del char_count[s[left]]
            left += 1
            
        # If window is size k and all characters are unique
        if right - left + 1 == k and len(char_count) == k:
            res.append(s[left:right+1])
            
    return res

s = "havefunonleetcode"
sub_strings(s, 5) 
```

# 4. Contains Duplicate II

Check if duplicate exists within window size k

Input: nums = [1,2,3,1], k = 3
Output: true
🟡 MEDIUM (VERY IMPORTANT)

```python
def contains_nearby_duplicate(nums, k):
    # Dictionary to store {number: last_index_seen}
    last_seen = {}
    
    for i, num in enumerate(nums):
        # If number was seen before and the distance is <= k
        if num in last_seen and i - last_seen[num] <= k:
            return True
        
        # Update the dictionary with the current index
        last_seen[num] = i
        
    return False

# Testing the example
nums = [1, 2, 3, 1]
k = 3
print(contains_nearby_duplicate(nums, k))  # Output: True
```
